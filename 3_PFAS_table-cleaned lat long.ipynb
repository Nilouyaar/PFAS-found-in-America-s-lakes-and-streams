```python
# I aggregate PFAS results to the monitoring-location level.
# This table is designed for both summary tables and map popups.
#
# Definitions:
#
# testing_instances
#     Number of PFAS result records at the location.
#Testing instances means number of PFAS result rows at that location. So if one site had 3 sampling dates and 40 PFAS chemicals tested each time, 
#testing instances could be 120.
#
# parameters_tested
#     Number of unique PFAS parameters tested at the location.
#
# parameters_positive
#     Number of unique PFAS parameters detected at least once.
#
# sampling_events
#     Number of unique sampling dates at the location.
#Sampling events means number of unique sampling dates at that location. So if the site was sampled on 3 different days, sampling events = 3
#
# detection_sampling_event_category
#     Whether PFAS was detected in 100% of sampling events or
#     in less than 100% of sampling events.
```

```python
file_path = "G:/My Drive/Projects/Large projects/PFAS in water/"

data_surface = pd.read_csv(
    file_path + "saved files/Notebook2/pfas_cleaned_analysis_file.csv",
    low_memory=False
)

print(data_surface.shape)
display(data_surface.head())
```

```python
# I check monitoring locations for missing latitude or longitude.

location_check = (
    data_surface[
        [
            "MonitoringLocationIdentifier",
            "MonitoringLocationName",
            "ActivityLocation/LatitudeMeasure",
            "ActivityLocation/LongitudeMeasure",
            "state",
            "OrganizationIdentifier",
            "OrganizationFormalName"
        ]
    ]
    .drop_duplicates()
    .copy()
)

location_check["missing_coordinates"] = (
    location_check["ActivityLocation/LatitudeMeasure"].isna()
    | location_check["ActivityLocation/LongitudeMeasure"].isna()
)

print(
    "Locations with missing coordinates:",
    location_check["missing_coordinates"].sum()
)

missing_coordinates = (
    location_check[
        location_check["missing_coordinates"]
    ]
    .copy()
    .sort_values(
        [
            "OrganizationIdentifier",
            "MonitoringLocationIdentifier"
        ]
    )
    .reset_index(drop=True)
)

print(
    "Total locations with missing coordinates:",
    len(missing_coordinates)
)

display(missing_coordinates)
```

```python
# I extract the USGS site number from the MonitoringLocationIdentifier
# for locations that are missing coordinates.

usgs_missing = missing_coordinates.copy()

usgs_missing["site_no"] = (
    usgs_missing["MonitoringLocationIdentifier"]
    .str.replace("USGS-", "", regex=False)
)

print(
    "USGS sites to retrieve:",
    len(usgs_missing)
)

display(
    usgs_missing[
        [
            "MonitoringLocationIdentifier",
            "site_no",
            "MonitoringLocationName",
            "OrganizationIdentifier"
        ]
    ]
)
```

```python
import time
import requests


# I create a list of the USGS site numbers for the
# monitoring locations with missing coordinates.

site_numbers = (
    usgs_missing["site_no"]
    .astype("string")
    .str.strip()
    .dropna()
    .unique()
    .tolist()
)


# I define the USGS site information API.

url = "https://waterservices.usgs.gov/nwis/site/"

params = {
    "format": "rdb",
    "sites": ",".join(site_numbers),
    "siteOutput": "expanded"
}


# I request the USGS site metadata and retry if the
# USGS web service is temporarily unavailable.

response = None

for attempt in range(5):
    try:
        response = requests.get(
            url,
            params=params,
            timeout=60
        )

        response.raise_for_status()

        print("USGS request successful.")
        break

    except requests.exceptions.RequestException as error:
        print(
            f"USGS request failed "
            f"(attempt {attempt + 1} of 5):",
            error
        )

        if attempt < 4:
            time.sleep(10)

else:
    raise RuntimeError(
        "USGS web service is currently unavailable. "
        "Try running this cell again later."
    )


print("Sites requested:", len(site_numbers))
```

```python
# I convert the successful USGS response into a dataframe.

from io import StringIO

# I remove USGS comment lines from the RDB response.

usgs_text = "\n".join(
    line
    for line in response.text.splitlines()
    if not line.startswith("#")
)

usgs_sites = pd.read_csv(
    StringIO(usgs_text),
    sep="\t",
    dtype="string"
)

# The first row contains RDB field definitions rather than
# monitoring-location data, so I remove it.

usgs_sites = usgs_sites.iloc[1:].copy()

usgs_sites["site_no"] = (
    usgs_sites["site_no"]
    .astype("string")
    .str.strip()
)

print(
    "Sites requested:",
    len(site_numbers)
)

print(
    "Sites returned:",
    usgs_sites["site_no"].nunique()
)

display(
    usgs_sites[
        [
            "site_no",
            "station_nm",
            "dec_lat_va",
            "dec_long_va",
            "state_cd"
        ]
    ]
)
```

```python
# I match the official USGS coordinates to the monitoring
# locations that are missing coordinates.

usgs_coordinate_check = (
    usgs_missing[
        [
            "MonitoringLocationIdentifier",
            "MonitoringLocationName",
            "site_no"
        ]
    ]
    .merge(
        usgs_sites[
            [
                "site_no",
                "station_nm",
                "dec_lat_va",
                "dec_long_va"
            ]
        ],
        on="site_no",
        how="left"
    )
)

# I convert the USGS coordinates to numeric values.

usgs_coordinate_check["dec_lat_va"] = pd.to_numeric(
    usgs_coordinate_check["dec_lat_va"],
    errors="coerce"
)

usgs_coordinate_check["dec_long_va"] = pd.to_numeric(
    usgs_coordinate_check["dec_long_va"],
    errors="coerce"
)

# I create lookup dictionaries for latitude and longitude.

latitude_lookup = (
    usgs_coordinate_check
    .set_index("MonitoringLocationIdentifier")["dec_lat_va"]
    .to_dict()
)

longitude_lookup = (
    usgs_coordinate_check
    .set_index("MonitoringLocationIdentifier")["dec_long_va"]
    .to_dict()
)

# I identify the USGS monitoring locations that need coordinates.

mask_usgs_sites = (
    data_surface["MonitoringLocationIdentifier"]
    .isin(
        usgs_coordinate_check[
            "MonitoringLocationIdentifier"
        ]
    )
)

# I identify rows with missing latitude.

mask_missing_lat = (
    mask_usgs_sites
    & data_surface[
        "ActivityLocation/LatitudeMeasure"
    ].isna()
)

# I fill the missing latitude values.

data_surface.loc[
    mask_missing_lat,
    "ActivityLocation/LatitudeMeasure"
] = (
    data_surface.loc[
        mask_missing_lat,
        "MonitoringLocationIdentifier"
    ]
    .map(latitude_lookup)
)

# I identify rows with missing longitude.

mask_missing_lon = (
    mask_usgs_sites
    & data_surface[
        "ActivityLocation/LongitudeMeasure"
    ].isna()
)

# I fill the missing longitude values.

data_surface.loc[
    mask_missing_lon,
    "ActivityLocation/LongitudeMeasure"
] = (
    data_surface.loc[
        mask_missing_lon,
        "MonitoringLocationIdentifier"
    ]
    .map(longitude_lookup)
)

print(
    "Rows with latitude filled:",
    mask_missing_lat.sum()
)

print(
    "Rows with longitude filled:",
    mask_missing_lon.sum()
)
```

```python
# I convert the USGS state FIPS codes for these locations
# to two-letter state abbreviations.

state_fips_to_abbr = {
    "06": "CA",
    "08": "CO",
    "26": "MI",
    "27": "MN",
    "34": "NJ",
    "35": "NM",
    "42": "PA",
    "55": "WI"
}


# I create a lookup between MonitoringLocationIdentifier
# and the state reported by USGS.

usgs_state_check = (
    usgs_missing[
        [
            "MonitoringLocationIdentifier",
            "site_no"
        ]
    ]
    .merge(
        usgs_sites[
            [
                "site_no",
                "state_cd"
            ]
        ],
        on="site_no",
        how="left"
    )
)


# I preserve leading zeros in the USGS state FIPS code.

usgs_state_check["state_cd"] = (
    usgs_state_check["state_cd"]
    .astype("string")
    .str.strip()
    .str.zfill(2)
)


# I convert the FIPS code to the state abbreviation.

usgs_state_check["state_usgs"] = (
    usgs_state_check["state_cd"]
    .map(state_fips_to_abbr)
)

display(usgs_state_check)
```

```python
# I create a lookup for the state of each USGS monitoring location.

state_lookup = (
    usgs_state_check
    .set_index(
        "MonitoringLocationIdentifier"
    )["state_usgs"]
    .to_dict()
)

# I identify these USGS locations where state is missing.

mask_missing_state = (
    data_surface["MonitoringLocationIdentifier"].isin(
        usgs_state_check[
            "MonitoringLocationIdentifier"
        ]
    )
    & data_surface["state"].isna()
)

# I fill the missing state values using the USGS metadata.

data_surface.loc[
    mask_missing_state,
    "state"
] = (
    data_surface.loc[
        mask_missing_state,
        "MonitoringLocationIdentifier"
    ]
    .map(state_lookup)
)

print(
    "Rows with state filled:",
    mask_missing_state.sum()
)
```

```python
# I verify that the 14 USGS locations now have state information.

check_14_usgs_state = (
    data_surface[
        data_surface["MonitoringLocationIdentifier"].isin(
            usgs_missing[
                "MonitoringLocationIdentifier"
            ]
        )
    ][
        [
            "MonitoringLocationIdentifier",
            "MonitoringLocationName",
            "state"
        ]
    ]
    .drop_duplicates()
    .sort_values("MonitoringLocationIdentifier")
    .reset_index(drop=True)
)

display(check_14_usgs_state)

print(
    "Unique locations:",
    check_14_usgs_state[
        "MonitoringLocationIdentifier"
    ].nunique()
)

print(
    "Missing state:",
    check_14_usgs_state[
        "state"
    ].isna().sum()
)
```

```python
# I aggregate PFAS results to the monitoring-location level.

#
# Definitions:
#
# testing_instances
# Total number of PFAS testing records at the location.
#
# parameters_tested
# Number of unique PFAS analytes tested at the location.
#
# parameters_positive
# Number of unique PFAS analytes classified as detected ("Y")
# at least once at the location.
#
# sampling_events
# Number of unique sampling dates at the location.
#
# sampling_events_with_detection
# Number of unique sampling dates when at least one PFAS analyte
# was classified as detected ("Y").
#
# max_pfas
# Highest reported numeric PFAS concentration at the location.


site_level = (
    data_surface
    .groupby(
        [
            "MonitoringLocationIdentifier",
            "MonitoringLocationName",
            "ActivityLocation/LatitudeMeasure",
            "ActivityLocation/LongitudeMeasure",
            "state"
        ],
        dropna=False
    )
    .agg(
        testing_instances=(
            "pfas_detected",
            "size"
        ),

        parameters_tested=(
            "pfas_analyte_standardized",
            "nunique"
        ),

        parameters_positive=(
            "pfas_analyte_standardized",
            lambda x: x[
                data_surface.loc[
                    x.index,
                    "pfas_detected"
                ].eq("Y")
            ].nunique()
        ),

        sampling_events=(
            "ActivityStartDate",
            "nunique"
        ),

        sampling_events_with_detection=(
            "pfas_detected",
            lambda x: (
                data_surface.loc[x.index]
                .groupby("ActivityStartDate")["pfas_detected"]
                .apply(
                    lambda y: y.eq("Y").any()
                )
                .sum()
            )
        ),

        max_pfas=(
            "ResultMeasureValue_num",
            "max"
        )
    )
    .reset_index()
)


# I calculate the percentage of sampling events at each location
# where at least one PFAS analyte was detected.

site_level["percent_sampling_events_with_detection"] = (
    site_level["sampling_events_with_detection"]
    / site_level["sampling_events"]
    * 100
)


# I classify locations based on the share of sampling events
# where at least one PFAS analyte was detected.

site_level["detection_sampling_event_category"] = np.select(
    [
        site_level[
            "percent_sampling_events_with_detection"
        ].eq(100),

        site_level[
            "percent_sampling_events_with_detection"
        ].gt(0)
        & site_level[
            "percent_sampling_events_with_detection"
        ].lt(100),

        site_level[
            "percent_sampling_events_with_detection"
        ].eq(0)
    ],
    [
        "PFAS detected in 100% of sampling events",
        "PFAS detected in less than 100% of sampling events",
        "No PFAS detected in sampling events"
    ],
    default="Unknown"
)
```

```python
# I identify the PFAS compound associated with the highest
# reported concentration at each monitoring location.

max_pfas_compound = (
    data_surface.loc[
        data_surface["ResultMeasureValue_num"].notna()
    ]
    .sort_values(
        "ResultMeasureValue_num",
        ascending=False
    )
    .drop_duplicates(
        subset="MonitoringLocationIdentifier",
        keep="first"
    )
    [
        [
            "MonitoringLocationIdentifier",
            "pfas_analyte_standardized",
            "ResultMeasureValue_num"
        ]
    ]
    .rename(
        columns={
            "pfas_analyte_standardized": "max_pfas_compound",
            "ResultMeasureValue_num": "max_pfas_check"
        }
    )
)

# I merge the compound associated with the maximum concentration
# into the site-level table.

site_level = site_level.merge(
    max_pfas_compound,
    on="MonitoringLocationIdentifier",
    how="left"
)

# I confirm that the concentration used to identify the compound
# matches the site-level maximum concentration.

print(
    "Rows where max concentration does not match:",
    (
        site_level["max_pfas"].notna()
        & ~site_level["max_pfas"].eq(
            site_level["max_pfas_check"]
        )
    ).sum()
)

display(
    site_level[
        [
            "MonitoringLocationName",
            "state",
            "max_pfas",
            "max_pfas_compound"
        ]
    ]
    .sort_values(
        "max_pfas",
        ascending=False
    )
    .head(20)
)

# The check column is no longer needed.

site_level = site_level.drop(
    columns="max_pfas_check"
)
```

```python
# I create a location-level PFAS detection flag.
#
# Y means at least one PFAS result was detected at the location.
# U means no PFAS result was detected, but at least one result was uncertain.
# N means all PFAS results at the location were classified as not detected.

site_detection_status = (
    data_surface
    .groupby(
        [
            "MonitoringLocationIdentifier",
            "MonitoringLocationName",
            "ActivityLocation/LatitudeMeasure",
            "ActivityLocation/LongitudeMeasure",
            "state"
        ],
        dropna=False
    )
    .agg(
        detected_results=(
            "pfas_detected",
            lambda x: x.eq("Y").sum()
        ),
        uncertain_results=(
            "pfas_detected",
            lambda x: x.eq("U").sum()
        ),
        not_detected_results=(
            "pfas_detected",
            lambda x: x.eq("N").sum()
        )
    )
    .reset_index()
)

site_level = site_level.merge(
    site_detection_status,
    on=[
        "MonitoringLocationIdentifier",
        "MonitoringLocationName",
        "ActivityLocation/LatitudeMeasure",
        "ActivityLocation/LongitudeMeasure",
        "state"
    ],
    how="left"
)

site_level["location_pfas_detected"] = np.select(
    [
        site_level["detected_results"] > 0,
        site_level["uncertain_results"] > 0,
        site_level["not_detected_results"] > 0
    ],
    [
        "Y",
        "U",
        "N"
    ],
    default="Unknown"
)

display(
    site_level["location_pfas_detected"]
    .value_counts(dropna=False)
)
```

```python
# I create a summary table based on location-level detection status.

location_detection_summary = (
    site_level
    .groupby(
        "location_pfas_detected",
        dropna=False
    )
    .agg(
        locations=(
            "MonitoringLocationIdentifier",
            "nunique"
        ),
        testing_instances=(
            "testing_instances",
            "sum"
        ),
        sampling_events=(
            "sampling_events",
            "sum"
        ),
        sampling_events_with_detection=(
            "sampling_events_with_detection",
            "sum"
        ),
        max_pfas=(
            "max_pfas",
            "max"
        )
    )
    .reset_index()
)


# I calculate the percentage of monitoring locations
# in each detection category.

location_detection_summary["percent_locations"] = (
    location_detection_summary["locations"]
    / site_level["MonitoringLocationIdentifier"].nunique()
    * 100
).round(1)


# I add the location-level detection status back to the
# record-level dataset so I can calculate unique PFAS analytes
# for each location detection category.

data_with_location_status = data_surface.merge(
    site_level[
        [
            "MonitoringLocationIdentifier",
            "location_pfas_detected"
        ]
    ].drop_duplicates(),
    on="MonitoringLocationIdentifier",
    how="left"
)


# I count each PFAS analyte only once within each
# location detection category.

parameter_summary = (
    data_with_location_status
    .groupby(
        "location_pfas_detected",
        dropna=False
    )
    .agg(
        parameters_tested=(
            "pfas_analyte_standardized",
            "nunique"
        )
    )
    .reset_index()
)


# I count each detected PFAS analyte only once within each
# location detection category.

positive_parameter_summary = (
    data_with_location_status.loc[
        data_with_location_status["pfas_detected"].eq("Y")
    ]
    .groupby(
        "location_pfas_detected",
        dropna=False
    )
    ["pfas_analyte_standardized"]
    .nunique()
    .reset_index(
        name="parameters_positive"
    )
)


# I combine the unique PFAS parameter counts with
# the location summary.

location_detection_summary = (
    location_detection_summary
    .merge(
        parameter_summary,
        on="location_pfas_detected",
        how="left"
    )
    .merge(
        positive_parameter_summary,
        on="location_pfas_detected",
        how="left"
    )
)

location_detection_summary["parameters_positive"] = (
    location_detection_summary["parameters_positive"]
    .fillna(0)
    .astype(int)
)

display(location_detection_summary)
```

```python
# I create a summary table based on sampling-level PFAS detection.
#
# This table separates locations into:
# - PFAS detected in 100% of sampling
# - PFAS detected in less than 100% of sampling
# - No PFAS detected

sampling_category_summary = (
    site_level
    .groupby(
        "detection_sampling_event_category",
        dropna=False
    )
    .agg(
        locations=(
            "MonitoringLocationIdentifier",
            "nunique"
        ),
        testing_instances=(
            "testing_instances",
            "sum"
        ),
        sampling_events=(
            "sampling_events",
            "sum"
        ),
        sampling_events_with_detection=(
            "sampling_events_with_detection",
            "sum"
        ),
        max_pfas=(
            "max_pfas",
            "max"
        )
    )
    .reset_index()
)


# I calculate the percentage of monitoring locations
# in each sampling detection category.

sampling_category_summary["percent_locations"] = (
    sampling_category_summary["locations"]
    / site_level["MonitoringLocationIdentifier"].nunique()
    * 100
).round(1)


# I add each location's sampling detection category back to
# the record-level dataset.

data_with_sampling_category = data_surface.merge(
    site_level[
        [
            "MonitoringLocationIdentifier",
            "detection_sampling_event_category"
        ]
    ].drop_duplicates(),
    on="MonitoringLocationIdentifier",
    how="left"
)


# I count each PFAS analyte only once within each
# sampling detection category.

sampling_parameter_summary = (
    data_with_sampling_category
    .groupby(
        "detection_sampling_event_category",
        dropna=False
    )
    .agg(
        parameters_tested=(
            "pfas_analyte_standardized",
            "nunique"
        )
    )
    .reset_index()
)


# I count each detected PFAS analyte only once within each
# sampling detection category.

sampling_positive_parameter_summary = (
    data_with_sampling_category.loc[
        data_with_sampling_category["pfas_detected"].eq("Y")
    ]
    .groupby(
        "detection_sampling_event_category",
        dropna=False
    )["pfas_analyte_standardized"]
    .nunique()
    .reset_index(
        name="parameters_positive"
    )
)


# I combine the unique PFAS parameter counts with
# the sampling category summary.

sampling_category_summary = (
    sampling_category_summary
    .merge(
        sampling_parameter_summary,
        on="detection_sampling_event_category",
        how="left"
    )
    .merge(
        sampling_positive_parameter_summary,
        on="detection_sampling_event_category",
        how="left"
    )
)


# I assign zero positive parameters to categories
# where no PFAS analytes were detected.

sampling_category_summary["parameters_positive"] = (
    sampling_category_summary["parameters_positive"]
    .fillna(0)
    .astype(int)
)


display(sampling_category_summary)
```

```python
# I highlighted three cases studies and I want to use their info
# I create a helper function to summarize PFAS monitoring

def summarize_case_study(case_name, location_mask):

    case_data = data_surface.loc[
        location_mask
    ].copy()

    # I identify the unique monitoring locations included
    # in the case study.

    case_locations = (
        case_data["MonitoringLocationIdentifier"]
        .dropna()
        .unique()
    )

    total_locations = len(case_locations)

    # I identify locations where at least one PFAS compound
    # was detected.

    detected_locations = (
        case_data.loc[
            case_data["pfas_detected"].eq("Y"),
            "MonitoringLocationIdentifier"
        ]
        .dropna()
        .nunique()
    )

    percent_locations_detected = (
        detected_locations / total_locations * 100
        if total_locations > 0
        else pd.NA
    )

    # I count the number of unique PFAS compounds detected
    # anywhere within the case-study area.

    compounds_detected = (
        case_data.loc[
            case_data["pfas_detected"].eq("Y"),
            "pfas_analyte_standardized"
        ]
        .dropna()
        .nunique()
    )

    # I identify the highest quantified PFAS concentration.
    # Only numeric concentration values are used for this metric.

    quantified_detected = case_data.loc[
        case_data["pfas_detected"].eq("Y")
        & case_data["ResultMeasureValue_num"].notna()
    ].copy()

    if len(quantified_detected) > 0:

        max_idx = (
            quantified_detected[
                "ResultMeasureValue_num"
            ]
            .idxmax()
        )

        highest_concentration = (
            quantified_detected.loc[
                max_idx,
                "ResultMeasureValue_num"
            ]
        )

        highest_compound = (
            quantified_detected.loc[
                max_idx,
                "pfas_analyte_standardized"
            ]
        )

        highest_location = (
            quantified_detected.loc[
                max_idx,
                "MonitoringLocationName"
            ]
        )

    else:
        highest_concentration = pd.NA
        highest_compound = pd.NA
        highest_location = pd.NA

    return {
        "case_study": case_name,
        "monitoring_locations": total_locations,
        "locations_with_pfas_detection": detected_locations,
        "percent_locations_with_detection": round(
            percent_locations_detected,
            1
        ),
        "pfas_compounds_detected": compounds_detected,
        "highest_pfas_ng_L": highest_concentration,
        "highest_pfas_compound": highest_compound,
        "highest_pfas_location": highest_location
    }
```

```python
# I create a state-by-state summary showing:
# - number of unique monitoring locations tested
# - number of locations where at least one PFAS was detected
# - percentage of locations where PFAS was detected
# - number of different PFAS compounds tested
# - number of different PFAS compounds detected


# Create one detection status for each monitoring location

# A location is considered to have PFAS if at least one
# record from that location has pfas_detected == "Y".

state_location_summary = (
    data_surface
    .groupby(
        ["state", "MonitoringLocationIdentifier"],
        dropna=False
    )
    .agg(
        location_pfas_detected=(
            "pfas_detected",
            lambda x: x.eq("Y").any()
        )
    )
    .reset_index()
)


# Count locations tested and locations with PFAS

state_location_summary = (
    state_location_summary
    .groupby("state", dropna=False)
    .agg(
        locations_tested=(
            "MonitoringLocationIdentifier",
            "nunique"
        ),
        locations_with_pfas=(
            "location_pfas_detected",
            "sum"
        )
    )
    .reset_index()
)


# Convert Boolean sum to integer.

state_location_summary[
    "locations_with_pfas"
] = (
    state_location_summary[
        "locations_with_pfas"
    ]
    .astype(int)
)


# Calculate the percentage of tested locations
# where at least one PFAS was detected.

state_location_summary[
    "percent_locations_with_pfas"
] = (
    state_location_summary[
        "locations_with_pfas"
    ]
    / state_location_summary[
        "locations_tested"
    ]
    * 100
).round(1)


#  Count different PFAS compounds tested in each state

state_compounds_tested = (
    data_surface
    .dropna(
        subset=["pfas_analyte_standardized"]
    )
    .groupby(
        "state",
        dropna=False
    )["pfas_analyte_standardized"]
    .nunique()
    .rename("pfas_compounds_tested")
    .reset_index()
)


#  Count different PFAS compounds detected in each state

state_compounds_found = (
    data_surface.loc[
        data_surface["pfas_detected"].eq("Y")
        & data_surface[
            "pfas_analyte_standardized"
        ].notna()
    ]
    .groupby(
        "state",
        dropna=False
    )["pfas_analyte_standardized"]
    .nunique()
    .rename("pfas_compounds_found")
    .reset_index()
)


#  Combine everything into the final state table

state_by_state_pfas = (
    state_location_summary
    .merge(
        state_compounds_tested,
        on="state",
        how="left"
    )
    .merge(
        state_compounds_found,
        on="state",
        how="left"
    )
)


# Replace missing compound counts with zero.

state_by_state_pfas[
    [
        "pfas_compounds_tested",
        "pfas_compounds_found"
    ]
] = (
    state_by_state_pfas[
        [
            "pfas_compounds_tested",
            "pfas_compounds_found"
        ]
    ]
    .fillna(0)
    .astype(int)
)


# Reorder and sort

state_by_state_pfas = (
    state_by_state_pfas[
        [
            "state",
            "locations_tested",
            "locations_with_pfas",
            "percent_locations_with_pfas",
            "pfas_compounds_tested",
            "pfas_compounds_found"
        ]
    ]
    .sort_values("state")
    .reset_index(drop=True)
)


display(state_by_state_pfas)


#  Save the final table

output_path = (
    file_path
    + "saved files/Notebook3/state_by_state_pfas_summary.csv"
)

state_by_state_pfas.to_csv(
    output_path,
    index=False
)

print("Saved to:", output_path)
```

```python
# ============================================================
# STATE-BY-STATE PFAS SUMMARY:
# 100% VS. LESS THAN 100% OF SAMPLING EVENTS
# ============================================================

# I classify each monitoring location based on the percentage
# of its sampling events in which at least one PFAS was detected.


# ------------------------------------------------------------
# 1. Create sampling-event detection status
# ------------------------------------------------------------

# A sampling event is defined by monitoring location and date.
# The event is positive if at least one PFAS was detected.

sampling_event_level = (
    data_surface
    .groupby(
        [
            "state",
            "MonitoringLocationIdentifier",
            "ActivityStartDate"
        ],
        dropna=False
    )
    .agg(
        event_pfas_detected=(
            "pfas_detected",
            lambda x: x.eq("Y").any()
        )
    )
    .reset_index()
)


# ------------------------------------------------------------
# 2. Calculate detection percentage for each location
# ------------------------------------------------------------

location_sampling_summary = (
    sampling_event_level
    .groupby(
        [
            "state",
            "MonitoringLocationIdentifier"
        ],
        dropna=False
    )
    .agg(
        sampling_events=(
            "event_pfas_detected",
            "size"
        ),
        sampling_events_with_pfas=(
            "event_pfas_detected",
            "sum"
        )
    )
    .reset_index()
)


location_sampling_summary[
    "percent_sampling_events_with_pfas"
] = (
    location_sampling_summary[
        "sampling_events_with_pfas"
    ]
    / location_sampling_summary[
        "sampling_events"
    ]
    * 100
)


# ------------------------------------------------------------
# 3. Classify locations
# ------------------------------------------------------------

location_sampling_summary[
    "detection_category"
] = "No PFAS detected"

location_sampling_summary.loc[
    location_sampling_summary[
        "percent_sampling_events_with_pfas"
    ].gt(0)
    & location_sampling_summary[
        "percent_sampling_events_with_pfas"
    ].lt(100),
    "detection_category"
] = "Less than 100%"

location_sampling_summary.loc[
    location_sampling_summary[
        "percent_sampling_events_with_pfas"
    ].eq(100),
    "detection_category"
] = "100%"


# ------------------------------------------------------------
# 4. Summarize locations by state
# ------------------------------------------------------------

state_detection_summary = (
    location_sampling_summary
    .groupby("state", dropna=False)
    .agg(
        locations_tested=(
            "MonitoringLocationIdentifier",
            "nunique"
        ),
        locations_100_percent=(
            "detection_category",
            lambda x: x.eq("100%").sum()
        ),
        locations_less_than_100_percent=(
            "detection_category",
            lambda x: x.eq("Less than 100%").sum()
        ),
        locations_no_pfas=(
            "detection_category",
            lambda x: x.eq("No PFAS detected").sum()
        )
    )
    .reset_index()
)


# Total locations where PFAS was found at least once

state_detection_summary[
    "locations_with_pfas"
] = (
    state_detection_summary["locations_100_percent"]
    + state_detection_summary[
        "locations_less_than_100_percent"
    ]
)


# Percentage of all tested locations where PFAS was found

state_detection_summary[
    "percent_locations_with_pfas"
] = (
    state_detection_summary["locations_with_pfas"]
    / state_detection_summary["locations_tested"]
    * 100
).round(1)


# ------------------------------------------------------------
# 5. Count standardized PFAS compounds tested
# ------------------------------------------------------------

state_compounds_tested = (
    data_surface
    .dropna(
        subset=["pfas_analyte_standardized"]
    )
    .groupby(
        "state",
        dropna=False
    )["pfas_analyte_standardized"]
    .nunique()
    .rename("pfas_compounds_tested")
    .reset_index()
)


# ------------------------------------------------------------
# 6. Count standardized PFAS compounds found
# ------------------------------------------------------------

state_compounds_found = (
    data_surface.loc[
        data_surface["pfas_detected"].eq("Y")
        & data_surface[
            "pfas_analyte_standardized"
        ].notna()
    ]
    .groupby(
        "state",
        dropna=False
    )["pfas_analyte_standardized"]
    .nunique()
    .rename("pfas_compounds_found")
    .reset_index()
)


# ------------------------------------------------------------
# 7. Combine everything
# ------------------------------------------------------------

state_by_state_100 = (
    state_detection_summary
    .merge(
        state_compounds_tested,
        on="state",
        how="left"
    )
    .merge(
        state_compounds_found,
        on="state",
        how="left"
    )
)


state_by_state_100[
    ["pfas_compounds_tested", "pfas_compounds_found"]
] = (
    state_by_state_100[
        ["pfas_compounds_tested", "pfas_compounds_found"]
    ]
    .fillna(0)
    .astype(int)
)


# ------------------------------------------------------------
# 8. Reorder columns
# ------------------------------------------------------------

state_by_state_100 = (
    state_by_state_100[
        [
            "state",
            "locations_tested",
            "locations_with_pfas",
            "percent_locations_with_pfas",
            "locations_100_percent",
            "locations_less_than_100_percent",
            "locations_no_pfas",
            "pfas_compounds_tested",
            "pfas_compounds_found"
        ]
    ]
    .sort_values("state")
    .reset_index(drop=True)
)


display(state_by_state_100)


# ------------------------------------------------------------
# 9. Save
# ------------------------------------------------------------

output_path = (
    file_path
    + "saved files/Notebook3/"
    + "state_by_state_pfas_100_percent_summary.csv"
)

state_by_state_100.to_csv(
    output_path,
    index=False
)

print("Saved to:", output_path)
```

```python

# I create a list of PFAS compounds detected at each
# individual monitoring location.
#
# A compound is considered detected if at least one
# record has pfas_detected == "Y".
#
# Each compound is counted only once per location,
# regardless of how many times it was detected.

# 1. Create a summary of all monitored locations.

location_pfas_summary = (
    data_surface
    .groupby(
        ["state", "MonitoringLocationIdentifier"],
        dropna=False
    )
    .agg(
        location_name=(
            "MonitoringLocationName",
            "first"
        ),
        pfas_compounds_tested=(
            "pfas_analyte_standardized",
            "nunique"
        )
    )
    .reset_index()
)

# 2. Keep only confirmed PFAS detections.

detected_records = data_surface.loc[
    data_surface["pfas_detected"].eq("Y")
    & data_surface["pfas_analyte_standardized"].notna()
].copy()

# 3. Create an alphabetical list of unique detected
# PFAS compounds for each monitoring location.

location_detected_compounds = (
    detected_records
    .groupby(
        ["state", "MonitoringLocationIdentifier"],
        dropna=False
    )
    .agg(
        pfas_compounds_found=(
            "pfas_analyte_standardized",
            "nunique"
        ),
        pfas_compound_names=(
            "pfas_analyte_standardized",
            lambda x: ", ".join(
                sorted(x.dropna().unique())
            )
        )
    )
    .reset_index()
)

# 4. Merge the detected compounds with all locations.
# This keeps locations where no PFAS was detected.

location_pfas_summary = (
    location_pfas_summary
    .merge(
        location_detected_compounds,
        on=["state", "MonitoringLocationIdentifier"],
        how="left",
        validate="one_to_one"
    )
)

# 5. Fill missing values for locations without detections.

location_pfas_summary["pfas_compounds_found"] = (
    location_pfas_summary["pfas_compounds_found"]
    .fillna(0)
    .astype(int)
)

location_pfas_summary["pfas_compound_names"] = (
    location_pfas_summary["pfas_compound_names"]
    .fillna("None detected")
)

# 6. Add a location-level detection status.

location_pfas_summary["pfas_detected"] = (
    location_pfas_summary["pfas_compounds_found"]
    .gt(0)
    .map({True: "Y", False: "N"})
)

# 7. Sort by state and monitoring location.

location_pfas_summary = (
    location_pfas_summary
    .sort_values(
        ["state", "MonitoringLocationIdentifier"]
    )
    .reset_index(drop=True)
)

# 8. Check that the number of locations matches
# the existing state-by-state summary.

location_count_check = (
    location_pfas_summary
    .groupby("state", dropna=False)
    .agg(
        locations_tested=(
            "MonitoringLocationIdentifier",
            "nunique"
        ),
        locations_with_pfas=(
            "pfas_detected",
            lambda x: x.eq("Y").sum()
        )
    )
    .reset_index()
)

comparison = location_count_check.merge(
    state_by_state_pfas[
        [
            "state",
            "locations_tested",
            "locations_with_pfas"
        ]
    ],
    on="state",
    suffixes=("_new", "_original"),
    validate="one_to_one"
)

mismatches = comparison.loc[
    (
        comparison["locations_tested_new"]
        != comparison["locations_tested_original"]
    )
    |
    (
        comparison["locations_with_pfas_new"]
        != comparison["locations_with_pfas_original"]
    )
]

print("States with mismatched location counts:", len(mismatches))
display(mismatches)

# 9. Display the final location-level table.

display(location_pfas_summary)

# 10. Save the results.

output_path = (
    file_path
    + "saved files/Notebook3/"
    + "pfas_compounds_by_monitoring_location.csv"
)

location_pfas_summary.to_csv(
    output_path,
    index=False
)

print("Saved to:", output_path)
```
